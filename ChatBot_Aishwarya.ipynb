```python
import google.generativeai as genai
from google.colab import userdata
```

```python
genai.configure(api_key=userdata.get('GEMINI_API_KEYS'))
```

```python
model = genai.GenerativeModel(
    model_name='gemini-2.5-flash-lite',
    system_instruction="""
    You are an Amazon Customer Support Assistant.

    Rules:
    - Be friendly and professional.
    - Give short and clear answers.
    - Help users with orders, returns, refunds, payments, delivery, and Prime membership.
    - If the question is unrelated to Amazon support, politely refuse.
    """
)
```

```python
knowledge_base = """
COMPANY: Amazon India Customer Support

==== ORDERS & DELIVERY ====
- Q: How can I track my order?
  A: Open Amazon app → Your Orders → Select the order → Track Package.
     You can see live delivery updates and estimated delivery time.

- Q: My package is delayed. What should I do?
  A: Check tracking updates in Your Orders. If the package is delayed beyond
     the expected date, contact Amazon support for help or compensation.

- Q: Can I cancel my order?
  A: Yes. Go to Your Orders → Select item → Cancel Item.
     Orders already shipped may not be cancellable.

- Q: I received the wrong item.
  A: Go to Your Orders → Return or Replace Items → Choose
     'Wrong item received' and request replacement or refund.

==== RETURNS & REFUNDS ====
- Q: How do I return a product?
  A: Open Your Orders → Select the product → Return or Replace.
     Choose reason and pickup option.

- Q: When will I get my refund?
  A: Refunds are usually processed within 3–7 business days after pickup
     or return verification.

- Q: Can I replace an item instead of refund?
  A: Yes, replacement is available for eligible products.

==== PAYMENTS ====
- Q: Which payment methods are accepted?
  A: Amazon accepts UPI, debit cards, credit cards, net banking,
     Amazon Pay balance, and cash on delivery for eligible orders.

- Q: My payment failed but money was deducted.
  A: Failed payment refunds are usually credited within 2–5 business days.

==== AMAZON PRIME ====
- Q: What are Prime benefits?
  A: Prime members get fast delivery, Prime Video, Prime Music,
     exclusive deals, and early sale access.

- Q: How can I cancel Prime membership?
  A: Go to Prime Membership settings → Manage Membership → End Membership.

==== ACCOUNT HELP ====
- Q: I forgot my password.
  A: Use the Forgot Password option on the login page to reset it.

- Q: My Amazon account is locked.
  A: Contact Amazon customer support to verify your identity and unlock the account.

==== GENERAL ====
- Q: Does Amazon provide 24/7 support?
  A: Yes, Amazon customer support is available 24/7 through chat and call support.
"""
```

```python
chat = model.start_chat(history=[])
```

```python

print("Hi, Im Amazon support assitant, How can I help you today?")

print("Bot is ready!, Type quit, bye , exit ")


while True:
  user_input = input("You: ")
  if user_input.lower() in ["quit", "bye", "exit"]:
    print("Bot: Goodbye!")
    break

  reply = chat.send_message(user_input, generation_config={"temperature": 0.2, "max_output_tokens": 200})
  print(f"Amazon Support: {reply.text}\n")
```
